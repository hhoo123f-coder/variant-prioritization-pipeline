# Target Gene Scoring Analysis

This notebook calculates the average prioritization score for each target gene (BRCA1, BRCA2, PALB2, TP53) based on the corrected prioritization results.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the corrected prioritization results file
df = pd.read_csv("prioritized_variants_CORRECTED.tsv", sep="\t")
df["Gene"] = df["GENEINFO"].str.split(":").str[0]

# Calculate total and average scores per gene
gene_scores = df.groupby("Gene").agg(
    Number_of_Variants=("ID", "count"),
    Total_Score=("Prioritization_Score", "sum"),
    Average_Gene_Score=("Prioritization_Score", "mean")
).reset_index()
gene_scores["Average_Gene_Score"] = gene_scores["Average_Gene_Score"].round(2)
gene_scores = gene_scores.sort_values(by="Average_Gene_Score", ascending=False)
display(gene_scores)
```

```python
# Create the average gene score chart
plt.figure(figsize=(9, 5))
bars = plt.bar(gene_scores["Gene"], gene_scores["Average_Gene_Score"])
plt.title("Average Target Gene Prioritization Score")
plt.xlabel("Target Gene")
plt.ylabel("Average Gene Score")
plt.bar_label(bars, fmt="%.2f", padding=4)
plt.ylim(0, gene_scores["Average_Gene_Score"].max() + 1)
plt.tight_layout()
plt.show()
```

```python
# Priority categories per gene
gene_categories = pd.crosstab(df["Gene"], df["Priority_Category"])
category_order = [
    "High educational priority",
    "Intermediate educational priority",
    "Low or uncertain educational priority"
]
gene_categories = gene_categories.reindex(columns=category_order, fill_value=0)
display(gene_categories)
```

```python
# Create stacked bar chart for priority categories
ax = gene_categories.plot(kind="bar", stacked=True, figsize=(10, 6))
ax.set_title("Variant Priority Categories by Target Gene")
ax.set_xlabel("Target Gene")
ax.set_ylabel("Number of Variants")
ax.legend(title="Priority Category")
plt.tight_layout()
plt.show()
```
